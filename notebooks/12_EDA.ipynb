## Data dictionary
|Name|Description|Type|Units
|--|--|--|--|
|cement|Cement (component 1)(kg in a m^3 mixture)|Float|Kg/m^3
|blast_furnance_slag|Blast Furnace Slag (component 2)(kg in a m^3 mixture)|Float|Kg/m^3|
|fly_ash|Fly Ash (component 3)(kg in a m^3 mixture)|Float|Kg/m^3
|water|Water (component 4)(kg in a m^3 mixture)|Float|Kg/m^3
|superplasticizer|Superplasticizer (component 5)(kg in a m^3 mixture)|Float|Kg/m^3|
|coarse_aggregate|Coarse Aggregate (component 6)(kg in a m^3 mixture)|Float|Kg/m^3
|fine_aggregate|Fine Aggregate (component 7)(kg in a m^3 mixture)|Float|Kg/m^3
|age|Age (day)|Int|day
|compresive_strength|Concrete compressive strength(MPa, megapascals)|Float|MPa

```python
df_baking = df_raw.copy()
df_baking.columns = (
    [
     'cement',
     'blast_furnance_slag',
     'fly_ash',
     'water',
     'superplasticizer',
     'coarse_aggregate',
     'fine_aggregate',
     'age',
     'compressive_strength'
     ]
)
df_baking['log_age'] = df_baking['age'].apply(lambda x: np.log(x))
df_baking = df_baking.drop(columns=['age'])
df = df_baking.copy()
df.info()
```

```python
df_train, df_test = train_test_split(df, test_size=0.1, random_state=2025)
df_train = df_train.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
df_train.head(5)
```

## **EDA for a categorical dependent variable**

```python
url = 'https://raw.githubusercontent.com/selva86/datasets/master/orange_juice_withmissing.csv'
df_juice_raw = pd.read_csv(url)
df_juice_raw.info()
```

```python
df_juice_raw.sample(10, random_state=2025)
```

```python
df_juice_baking = df_juice_raw.copy()
df_juice_baking = df_juice_baking.drop(columns=['STORE', 'Store7', 'SalePriceMM', 'SalePriceCH', 'ListPriceDiff', 'PctDiscMM', 'PctDiscCH'])
df_juice_baking.columns = df_juice_baking.columns.str.lower()
df_juice_baking['purchase'] = df_juice_baking['purchase'].astype('category')
df_juice_baking['storeid'] = df_juice_baking['storeid'].astype('category')
df_juice_baking['specialch'] = df_juice_baking['specialch'].astype('category')
df_juice_baking['specialmm'] = df_juice_baking['specialmm'].astype('category')
df_juice_baking = df_juice_baking.dropna()
df_juice_baking['storeid'] = df_juice_baking['storeid'].apply(lambda x: 'store_'+str(int(x)))
df_juice_baking['specialch'] = df_juice_baking['specialch'].apply(lambda x: 'spch_'+str(int(x)))
df_juice_baking['specialmm'] = df_juice_baking['specialmm'].apply(lambda x: 'spmm_'+str(int(x)))
df_juice = df_juice_baking.copy()
df_juice.info()
```

```python
df_juice_train, df_juice_test = train_test_split(
    df_juice, test_size=0.1, stratify=df_juice['purchase']
    )
df_juice_train = df_juice_train.reset_index(drop=True)
df_juice_test = df_juice_test.reset_index(drop=True)
df_juice_train.shape
```

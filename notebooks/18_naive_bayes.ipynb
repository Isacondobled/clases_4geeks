```python
#Step 0: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, f1_score, r2_score
import numpy as np
from sklearn.model_selection import RandomizedSearchCV
```

```python
#Step 1: Load data
url = 'https://raw.githubusercontent.com/4GeeksAcademy/naive-bayes-project-tutorial/main/playstore_reviews.csv'
df_raw = pd.read_csv(url)
df_raw
```

```python
df_baking = df_raw.copy()
df_baking = df_baking.drop(columns='package_name')
df = df_baking.copy()
```

```python
df_train, df_test = train_test_split(df, test_size=0.1, random_state=2025)
df_train = df_train.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
#Step 2: Machine Learning
X_train = df['review']
y_train = df['polarity']

X_test = df['review']
y_test = df['polarity']
```

```python
vec_model = CountVectorizer(stop_words='english')
X_train = vec_model.fit_transform(X_train).toarray()
X_test = vec_model.transform(X_test).toarray()
```

```python
nb = GaussianNB()
nb.fit(X_train, y_train)

nbm = MultinomialNB()
nbm.fit(X_train, y_train)

nbb = BernoulliNB()
nbb.fit(X_train, y_train)
```

```python
y_hat = nb.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, y_hat):.2f}, F1 score: {f1_score(y_test, y_hat):.2f}, R2 score: {r2_score(y_test, y_hat):.2f}')
```

```python
y_hat = nbm.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, y_hat):.2f}, F1 score: {f1_score(y_test, y_hat):.2f}, R2 score: {r2_score(y_test, y_hat):.2f}')
```

```python
y_hat = nbb.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, y_hat):.2f}, F1 score: {f1_score(y_test, y_hat):.2f}, R2 score: {r2_score(y_test, y_hat):.2f}')
```

```python

hyperparams = {
    'var_smoothing': np.logspace(0, -9, num=100)
}


random_search = RandomizedSearchCV(nb, hyperparams, n_iter = 50, scoring = "accuracy", cv = 5, random_state = 42)
random_search
```

```python
random_search.fit(X_train, y_train)
```

```python
y_hat = random_search.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, y_hat):.2f}, F1 score: {f1_score(y_test, y_hat):.2f}, R2 score: {r2_score(y_test, y_hat):.2f}')
```

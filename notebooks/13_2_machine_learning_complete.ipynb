### 1. **Data loading**

```python
#Step 1 Load data
url = 'https://raw.githubusercontent.com/selva86/' + \
'datasets/master/orange_juice_withmissing.csv'
df_raw = pd.read_csv(url)
df_raw.info()
```

### 2. **Data preprocessing**
**Objectives**: Perform the data cleaning, data transformation and data reduction steps to avoid data mistmatching, noisy data or data not wrangled

```python
#Step 2 Preprocessing
df_baking = df_raw.copy()
df_baking = df_baking.drop(columns=['STORE', 'Store7', 'SalePriceMM', 'SalePriceCH', 'ListPriceDiff', 'PctDiscMM', 'PctDiscCH'])
df_baking.columns = df_baking.columns.str.lower()
df_baking['purchase'] = df_baking['purchase'].astype('category')
df_baking['storeid'] = df_baking['storeid'].astype('category')
df_baking['specialch'] = df_baking['specialch'].astype('category')
df_baking['specialmm'] = df_baking['specialmm'].astype('category')
df_baking = df_baking.dropna()
df_baking['storeid'] = df_baking['storeid'].apply(lambda x: 'store_'+str(int(x)))
df_baking['specialch'] = df_baking['specialch'].apply(lambda x: 'spch_'+str(int(x)))
df_baking['specialmm'] = df_baking['specialmm'].apply(lambda x: 'spmm_'+str(int(x)))
df = df_baking.copy()
df.info()
```

### 4. **Machine learning**
**Objective**: Create a model that learns from data to make predictions and generalize to unseen data, and thus perform tasks without explicit instructions

```python
#Step 4 ML
X_train = df_train.drop(columns=['purchase'])
y_train = df_train['purchase']
X_val = df_val.drop(columns=['purchase'])
y_val = df_val['purchase']
X_test = df_test.drop(columns=['purchase'])
y_test = df_test['purchase']
```

```python
label_encoder = LabelEncoder()
y_train_enc = label_encoder.fit_transform(y_train)
y_val_enc = label_encoder.transform(y_val)
y_test_enc = label_encoder.transform(y_test)
```

```python
num_cols = X_train.select_dtypes('number').columns
cat_cols = X_train.select_dtypes('category').columns

num_proc = Pipeline(steps=[
    ('scaler', MinMaxScaler())
])
cat_proc = Pipeline(steps=[
    ('onehot', OneHotEncoder(drop='first'))
])
processor = ColumnTransformer(transformers=[
    ('num', num_proc, num_cols),
    ('proc', cat_proc, cat_cols)
])
```

```python
clf_tree = Pipeline([
    ('proc', processor),
    ('tree', DecisionTreeClassifier(max_depth=6,
                                    random_state=2025,
                                    class_weight='balanced'))
])

clf_lr = Pipeline([
    ('proc', processor),
    ('lreg', LogisticRegression(random_state=2025,
                                  class_weight='balanced'))
])

clf_rf = Pipeline([
    ('proc', processor),
    ('rf', RandomForestClassifier(random_state=2025,
                                  class_weight='balanced'))
])

clf_hb = Pipeline([
    ('proc', processor),
    ('hb', HistGradientBoostingClassifier(random_state=2025,
                                          class_weight='balanced'))
])

clf_nb = Pipeline([
    ('proc', processor),
    ('nb', GaussianNB())
])
```

```python
clf_tree.fit(X_train, y_train_enc)
y_hat = clf_tree.predict(X_val)
print(f'Accuracy: {accuracy_score(y_val_enc, y_hat):.2f}, F1 score: {f1_score(y_val_enc, y_hat):.2f}, R2 score: {r2_score(y_val_enc, y_hat):.2f}')
```

```python
models = [
          (clf_lr,'Logistic Regression', 'lreg__'),
          (clf_tree,'Decission Tree', 'tree__'),
          (clf_nb,'Naive Bayes', 'nb__'),
          (clf_rf,'Random Forest', 'rf__'),
          (clf_hb,'Histogram GBoosting', 'hb__')
]
performance = {}
for est,name in models:
  est.fit(X_train, y_train_enc)
  y_hat = est.predict(X_val)
  acc = round(accuracy_score(y_val_enc, y_hat), 2)
  f1 = round(f1_score(y_val_enc, y_hat), 2)
  r2 = round(r2_score(y_val_enc, y_hat), 2)
  performance[name] = {
      'accuracy':acc,
      'F1 Score':f1,
      'R2 Score':r2
  }
```

```python
param_grids = {
    'lreg':{
        'lreg__C':[0.01, 0.1, 1]
    },
    'tree':{
        'tree__max_depth':[2,4,5,6,7],
        'tree__min_samples_split':[2,4,6],
    },
    'nb':{
        'nb__var_smoothing':[1e9, 1e-8]
    },
    'rf':{
        'rf__max_depth':[2,4,5,6,7],
        'rf__min_samples_split':[2,4,6],
        'rf__n_estimators':[50,100,150,200]
    },
    'hb':{
        'hb__max_depth':[2,4,5,6,7],
    }
}

models = [
          (clf_lr,'Logistic Regression', 'lreg'),
          (clf_tree,'Decission Tree', 'tree'),
          (clf_nb,'Naive Bayes', 'nb'),
          (clf_rf,'Random Forest', 'rf'),
          (clf_hb,'Histogram GBoosting', 'hb')
]

performance = {}
for est, name, sname in models:
  est.fit(X_train, y_train_enc)
  estimator_cv = GridSearchCV(
      est,
      param_grid = param_grids[sname],
      cv = 5
  )
  estimator_cv.fit(X_train, y_train_enc)
  y_hat = estimator_cv.predict(X_val)
  acc = round(accuracy_score(y_val_enc, y_hat), 2)
  f1 = round(f1_score(y_val_enc, y_hat), 2)
  r2 = round(r2_score(y_val_enc, y_hat), 2)
  best_params = estimator_cv.best_params_
  performance[name] = {
      'accuracy':acc,
      'F1 Score':f1,
      'R2 Score':r2,
      'Best Params': best_params,
      'estimator': estimator_cv.best_estimator_
  }
```

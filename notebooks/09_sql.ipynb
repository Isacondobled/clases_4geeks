```python
from pandasql import sqldf
import os
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/syntagmatic/gapminder-csv/refs/heads/master/gapminder.csv"
df = pd.read_csv(url)
```

```python
df.info()
```

```python
sqldf("SELECT * FROM df LIMIT 5") #children_per_woman_total_fertility life_expetancy_years
```

```python
#sqldf("SELECT children_per_woman_total_fertility, life_expectancy_years FROM df LIMIT 10")
sqldf("SELECT geo, time, children_per_woman_total_fertility, life_expectancy_years FROM df LIMIT 10")
```

```python
sqldf("SELECT geo, time, children_per_woman_total_fertility, life_expectancy_years FROM df WHERE geo='arg' and time>1975 LIMIT 10")
```

```python
sqldf("SELECT geo, time, children_per_woman_total_fertility, life_expectancy_years FROM df WHERE geo IN ('col','arg') and time>2000")
```

```python
sqldf("SELECT geo, time, children_per_woman_total_fertility, life_expectancy_years FROM df WHERE geo IN ('col','arg') AND time = 2000")
```

```python
sqldf("SELECT geo, time, children_per_woman_total_fertility, life_expectancy_years FROM df WHERE geo IN ('col','arg','mex') AND time = 2000")
```

```python
sqldf("SELECT DISTINCT geo FROM df")
```

```python
sqldf("SELECT geo, gdp_total_yearly_growth FROM df WHERE geo='per' AND time BETWEEN 1985 AND 2000 ")
```

```python
sqldf("SELECT geo, time, gdp_total_yearly_growth FROM df WHERE geo='per' AND time BETWEEN 1985 AND 2000 ORDER BY time DESC")
```

```python
sqldf("SELECT geo, time, gdp_total_yearly_growth FROM df WHERE geo='arg' AND time BETWEEN 1990 AND 2010 ORDER BY time DESC")
```

```python
sqldf("SELECT time, MAX(gdp_total_yearly_growth) AS max_gdp FROM df WHERE geo='mex'")
```

```python
sqldf("SELECT MAX(gdp_total_yearly_growth), geo FROM df GROUP BY geo LIMIT 20")
```

```python
sqldf("SELECT MAX(gdp_total_yearly_growth), geo, time FROM df WHERE geo IN ('mex', 'arg', 'per') AND time>1985 AND 2025 GROUP BY geo")
```

```python
sqldf("SELECT MIN(gdp_total_yearly_growth), geo, time FROM df WHERE geo IN ('mex', 'arg', 'per') AND time>1985 AND 2025 GROUP BY geo")
```

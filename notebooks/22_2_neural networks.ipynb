```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def sigmoid(x: float):
  return 1/(1+np.exp(-x))

x = np.linspace(-10, 10, 200)
y = sigmoid(x)
```

```python
plt.plot(x, y, label='sigmoid function')
plt.legend()
plt.grid(True)
plt.show()
```

```python
sigmoid(0)
```

```python
def derivative_sigmoide(x:float):
  return sigmoid(x)*(1-sigmoid(x))

y_diff = derivative_sigmoide(x)
```

```python
plt.plot(x, y, label='sigmoid function')
plt.plot(x, y_diff, label='derivative sigmoide function')
plt.legend()
plt.grid(True)
plt.show()
```
